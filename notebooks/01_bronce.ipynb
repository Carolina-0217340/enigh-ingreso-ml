```python
from pyspark.sql import functions as F

base = "/Volumes/workspace/enigh/datos/"
f_conc = [f.path for f in dbutils.fs.ls(base) if "concentrado" in f.name.lower()][0]
f_gast = [f.path for f in dbutils.fs.ls(base) if "gastos" in f.name.lower()][0]

def a_bronce(ruta, tabla):
    df = (spark.read.option("header", True).option("inferSchema", False).csv(ruta)
          .withColumn("_fecha_ingesta", F.current_timestamp())
          .withColumn("_archivo_origen", F.lit(ruta.split("/")[-1])))
    df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(f"workspace.enigh.{tabla}")
    print(tabla, "->", spark.table(f"workspace.enigh.{tabla}").count(), "filas")

a_bronce(f_conc, "bronce_concentrado")
a_bronce(f_gast, "bronce_gastos")
```
